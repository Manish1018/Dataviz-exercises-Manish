# Analytical Question 1

## How has Netflix content grown over the years?

```python
yearly_content = df.groupby('release_year').size().reset_index(name='Total Titles')

yearly_content.tail()
```

```python
fig = px.line(
    yearly_content,
    x='release_year',
    y='Total Titles',
    title='Growth of Netflix Content Over the Years',
    markers=True
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Release Year',
    yaxis_title='Number of Titles'
)

fig.show()
```

```python
fig = px.line(
    yearly_content,
    x='release_year',
    y='Total Titles',
    title='Growth of Netflix Content Over the Years',
    markers=True
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Release Year',
    yaxis_title='Number of Titles'
)

fig.show()
```

### Insight

Netflix's content library expanded significantly after 2015, reaching its highest levels around 2018–2020. This indicates a rapid increase in content production and acquisition during those years.

# Analytical Question 2

## Which countries contribute the most Netflix titles?

```python
country_data = (
    df[df['country'] != 'Unknown']
    .assign(country=df['country'].str.split(', '))
    .explode('country')
)

top_countries = (
    country_data['country']
    .value_counts()
    .head(10)
    .reset_index()
)

top_countries.columns = ['Country', 'Titles']

top_countries
```

```python
fig = px.bar(
    top_countries,
    x='Country',
    y='Titles',
    color='Titles',
    title='Top 10 Countries Producing Netflix Content'
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Country',
    yaxis_title='Number of Titles'
)

fig.show()    
```

### Insight

The United States contributes the largest number of Netflix titles, followed by countries such as India and the United Kingdom. Netflix content production is concentrated in a relatively small number of countries.

# Analytical Question 3

## What are the most popular genres on Netflix?

```python
genre_data = (
    df.assign(Genre=df['listed_in'].str.split(', '))
    .explode('Genre')
)

top_genres = (
    genre_data['Genre']
    .value_counts()
    .head(10)
    .reset_index()
)

top_genres.columns = ['Genre', 'Titles']
```

```python
fig = px.bar(
    top_genres,
    x='Genre',
    y='Titles',
    color='Titles',
    title='Top 10 Netflix Genres'
)

fig.update_layout(
    template='plotly_white'
)

fig.show()
```

### Insight

International Movies, Dramas, and Comedies are among the most common genres on Netflix, reflecting the platform's diverse global content library.

# Analytical Question 4

## How do Movies and TV Shows compare across content ratings?

```python
rating_type = (
    df.groupby(['rating', 'type'])
    .size()
    .reset_index(name='Titles')
)

rating_type.head()
```

```python
fig = px.bar(
    rating_type,
    x='rating',
    y='Titles',
    color='type',
    barmode='group',
    title='Comparison of Movies and TV Shows Across Ratings'
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Content Rating',
    yaxis_title='Number of Titles'
)

fig.show()
```

### Insight

TV-MA and TV-14 dominate Netflix's catalog. Movies and TV Shows have different rating distributions, showing that Netflix targets multiple audience age groups.

# Analytical Question 5

## Which directors have produced the most Netflix titles?

```python
director_data = (
    df[df['director'] != 'Unknown']
    .assign(director=df['director'].str.split(', '))
    .explode('director')
)

top_directors = (
    director_data['director']
    .value_counts()
    .head(10)
    .reset_index()
)

top_directors.columns = ['Director', 'Titles']
```

```python
fig = px.bar(
    top_directors,
    x='Director',
    y='Titles',
    color='Titles',
    title='Top 10 Directors on Netflix'
)

fig.update_layout(
    template='plotly_white'
)

fig.show()
```

### Insight

A small number of directors are responsible for multiple Netflix titles, while most directors appear only once or twice, indicating a diverse creator base.

# Analytical Question 6

## Which months see the highest number of titles added to Netflix?

```python
month_order = [
    'January','February','March','April','May','June',
    'July','August','September','October','November','December'
]

month_data = (
    df['date_added_month']
    .value_counts()
    .reindex(month_order)
    .reset_index()
)

month_data.columns = ['Month', 'Titles']
```

```python
fig = px.line(
    month_data,
    x='Month',
    y='Titles',
    markers=True,
    title='Titles Added to Netflix by Month'
)

fig.update_layout(
    template='plotly_white'
)

fig.show()
```

### Insight

Netflix tends to add more titles during certain months, suggesting seasonal content release strategies throughout the year.

# Analytical Question 7

## Which release years have the highest average movie duration?

```python
movie_duration = (
    df[df['duration_unit'] == 'min']
    .groupby('release_year')['duration_value']
    .mean()
    .reset_index()
)

movie_duration.columns = ['Release Year', 'Average Duration']
```

```python
fig = px.line(
    movie_duration,
    x='Release Year',
    y='Average Duration',
    markers=True,
    title='Average Movie Duration by Release Year'
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5
)

fig.show()
```

### Insight

The average duration of movies has remained relatively stable over the years, with only small fluctuations between release years.

# Analytical Question 8

## Which content ratings are most common in each content type?

```python
rating_content = (
    df.groupby(['type', 'rating'])
    .size()
    .reset_index(name='Titles')
)
```

```python
fig = px.sunburst(
    rating_content,
    path=['type', 'rating'],
    values='Titles',
    title='Content Ratings by Movies and TV Shows'
)

fig.update_layout(
    template='plotly_white'
)

fig.show()
```

### Insight

Movies and TV Shows differ in their content ratings, with TV-MA and TV-14 representing a large share of Netflix's library.

# Analytical Question 9

## Which countries produce the greatest variety of Netflix genres?

```python
country_genre = (
    df[df['country'] != 'Unknown']
    .assign(country=df['country'].str.split(', '))
    .explode('country')
    .assign(Genre=lambda x: x['listed_in'].str.split(', '))
    .explode('Genre')
)

genre_diversity = (
    country_genre.groupby('country')['Genre']
    .nunique()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

genre_diversity.columns = ['Country', 'Unique Genres']
```

```python
fig = px.bar(
    genre_diversity,
    x='Country',
    y='Unique Genres',
    color='Unique Genres',
    title='Countries with the Highest Genre Diversity'
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5
)

fig.show()
```

### Insight

Countries such as the United States and India produce content across a wide range of genres, demonstrating a highly diverse entertainment industry.

# Analytical Question 10

## How has the balance between Movies and TV Shows changed over time?

```python
type_year = (
    df.groupby(['release_year', 'type'])
    .size()
    .reset_index(name='Titles')
)
```

```python
fig = px.area(
    type_year,
    x='release_year',
    y='Titles',
    color='type',
    title='Movies and TV Shows Released Over Time'
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5
)

fig.show()
```

### Insight

Movies consistently make up the majority of Netflix's catalog, while the number of TV Shows has steadily increased over recent years.
